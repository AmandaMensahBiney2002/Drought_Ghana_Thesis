```python
import os
import rioxarray
import geopandas as gpd
from tqdm import tqdm

# Load Ghana boundary
ghana = gpd.read_file('/content/drive/MyDrive/Drought_Ghana_Thesis/data/shapefiles/gadm41_GHA_0.shp')

# Input and output folders
chirps_dir = '/content/drive/MyDrive/Drought_Ghana_Thesis/data/raw/chirps'
output_folder = '/content/drive/MyDrive/Drought_Ghana_Thesis/data/processed/ghana_monthly'

# Create output folder
os.makedirs(output_folder, exist_ok=True)

# Get all files
tif_files = sorted([f for f in os.listdir(chirps_dir) if f.endswith('.tif')])

# Track progress
failed = []
skipped = 0
clipped = 0

for filename in tqdm(tif_files, desc="Clipping to Ghana"):
    input_file = os.path.join(chirps_dir, filename)
    output_file = os.path.join(output_folder, filename)

    # Skip if already clipped
    if os.path.exists(output_file):
        skipped += 1
        continue

    try:
        # Open
        chirps = rioxarray.open_rasterio(input_file)

        # Ensure CRS
        if chirps.rio.crs is None:
            chirps.rio.write_crs("EPSG:4326", inplace=True)

        # Clip to Ghana
        chirps_ghana = chirps.rio.clip(
            ghana.geometry,
            ghana.crs,
            drop=True,
            all_touched=True
        )

        # Save
        chirps_ghana.rio.to_raster(output_file)

        # Close
        chirps.close()
        chirps_ghana.close()
        clipped += 1

    except Exception as e:
        failed.append((filename, str(e)))
        print(f"\n❌ Failed: {filename} — {e}")

print(f"\n\n{'=' * 50}")
print(f"CLIPPING COMPLETE")
print(f"{'=' * 50}")
print(f"Total files: {len(tif_files)}")
print(f"Clipped: {clipped}")
print(f"Skipped (already done): {skipped}")
print(f"Failed: {len(failed)}")
```

```python
import os
import rioxarray
import geopandas as gpd
from tqdm import tqdm
import time

# Load Ghana boundary
ghana = gpd.read_file('/content/drive/MyDrive/Drought_Ghana_Thesis/data/shapefiles/gadm41_GHA_0.shp')

chirps_dir = '/content/drive/MyDrive/Drought_Ghana_Thesis/data/raw/chirps'
output_folder = '/content/drive/MyDrive/Drought_Ghana_Thesis/data/processed/ghana_monthly'

# Get remaining files
raw_files = sorted([f for f in os.listdir(chirps_dir) if f.endswith('.tif')])
clipped_existing = set(os.listdir(output_folder))
remaining = [f for f in raw_files if f not in clipped_existing]

# Process in batches of 50
BATCH_SIZE = 50
failed = []

for batch_num in range(0, len(remaining), BATCH_SIZE):
    batch = remaining[batch_num:batch_num + BATCH_SIZE]
    print(f"\n{'=' * 50}")
    print(f"BATCH {batch_num // BATCH_SIZE + 1}")
    print(f"Processing files {batch_num + 1} to {batch_num + len(batch)}")
    print(f"{'=' * 50}")

    batch_failed = 0

    for filename in tqdm(batch, desc=f"Batch {batch_num // BATCH_SIZE + 1}"):
        input_file = os.path.join(chirps_dir, filename)
        output_file = os.path.join(output_folder, filename)

        if os.path.exists(output_file):
            continue

        try:
            chirps = rioxarray.open_rasterio(input_file)

            if chirps.rio.crs is None:
                chirps.rio.write_crs("EPSG:4326", inplace=True)

            chirps_ghana = chirps.rio.clip(
                ghana.geometry,
                ghana.crs,
                drop=True,
                all_touched=True
            )

            chirps_ghana.rio.to_raster(output_file)

            chirps.close()
            chirps_ghana.close()

        except Exception as e:
            failed.append((filename, str(e)))
            batch_failed += 1
            print(f"\n❌ Failed: {filename} — {e}")

    print(f"\n✅ Batch {batch_num // BATCH_SIZE + 1} complete")
    print(f"   Failures in this batch: {batch_failed}")

    # Short pause between batches to let the connection recover
    if batch_num + BATCH_SIZE < len(remaining):
        print(f"   Pausing 5 seconds before next batch...")
        time.sleep(5)

print(f"\n\n{'=' * 50}")
print(f"ALL BATCHES COMPLETE")
print(f"{'=' * 50}")
print(f"Total failures: {len(failed)}")
```

```python
import os
import urllib.request
from tqdm import tqdm

# Base URL for CHIRPS v3.0 Africa monthly TIFFs
base_url = "https://data.chc.ucsb.edu/products/CHIRP-v3.0/monthly/africa/tifs"

# Local re-download folder (in Colab's fast local storage, not Google Drive)
redownload_dir = '/content/redownloaded'
os.makedirs(redownload_dir, exist_ok=True)

corrupted = [
    'chirp-v3.0.1989.02.tif',
    'chirp-v3.0.1989.11.tif',
    'chirp-v3.0.1990.01.tif',
    'chirp-v3.0.1990.03.tif',
    'chirp-v3.0.1990.05.tif',
    'chirp-v3.0.2000.09.tif',
    'chirp-v3.0.2003.12.tif',
    'chirp-v3.0.2008.11.tif',
    'chirp-v3.0.2009.08.tif',
    'chirp-v3.0.2016.06.tif',
    'chirp-v3.0.2016.07.tif',
]

print("Re-downloading corrupted files...")

for filename in tqdm(corrupted, desc="Downloading"):
    url = f"{base_url}/{filename}"
    local_path = os.path.join(redownload_dir, filename)

    try:
        urllib.request.urlretrieve(url, local_path)
    except Exception as e:
        print(f"\n❌ Failed: {filename} — {e}")

print(f"\n✅ Re-download complete")
print(f"Files in {redownload_dir}: {len(os.listdir(redownload_dir))}")
```
